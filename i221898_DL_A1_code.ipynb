```python
# ============================================================
# Enhanced Multi-Task Emotion Recognition with Fixed Custom CNN
# ============================================================

import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import (accuracy_score, f1_score, cohen_kappa_score, 
                           roc_auc_score, average_precision_score, 
                           confusion_matrix, classification_report)
from sklearn.preprocessing import label_binarize
import pandas as pd
from scipy.stats import pearsonr

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
import torchvision.models as models
from torch.optim.lr_scheduler import ReduceLROnPlateau
import torch.nn.functional as F


# ============================================================
# Dataset Loader (Enhanced with better augmentation)
# ============================================================
class MultiTaskDataset(Dataset):
    def __init__(self, images_dir: str, annotations_dir: str, img_size=(224,224),
                 augment=False, split="train", split_ratio=0.8, seed=42):
        self.images_dir = images_dir
        self.annotations_dir = annotations_dir
        self.img_size = img_size
        self.augment = augment

        all_samples = self._gather_samples()
        random.seed(seed)
        random.shuffle(all_samples)
        
        # Handle three-way split: train/val/test
        if split == "train":
            split_idx = int(len(all_samples) * split_ratio)
            self.samples = all_samples[:split_idx]
        elif split == "val":
            train_split = int(len(all_samples) * 0.7)  # 70% train
            val_split = int(len(all_samples) * 0.85)   # 15% val (70% + 15%)
            self.samples = all_samples[train_split:val_split]
        else:  # test
            test_split = int(len(all_samples) * 0.85)  # 15% test (remaining)
            self.samples = all_samples[test_split:]

        # Enhanced transforms with better augmentation
        base_tfms = [
            T.Resize(self.img_size),
            T.ToTensor(),
            T.Normalize(mean=[0.485, 0.456, 0.406],
                        std=[0.229, 0.224, 0.225])
        ]

        if self.augment and split == "train":
            self.transform = T.Compose([
                T.Resize((int(self.img_size[0] * 1.1), int(self.img_size[1] * 1.1))),
                T.RandomCrop(self.img_size),
                T.RandomHorizontalFlip(p=0.5),
                T.RandomRotation(15),
                T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05),
                T.RandomGrayscale(p=0.05),
                T.ToTensor(),
                T.Normalize(mean=[0.485, 0.456, 0.406],
                            std=[0.229, 0.224, 0.225])
            ])
        else:
            self.transform = T.Compose(base_tfms)
            
        # Store original transform for visualization
        self.viz_transform = T.Compose([
            T.Resize(self.img_size),
            T.ToTensor(),
        ])

    def _gather_samples(self):
        samples = []
        files = os.listdir(self.annotations_dir)
        prefixes = sorted(set(f.split("_")[0] for f in files))

        for prefix in prefixes:
            exp_file = os.path.join(self.annotations_dir, f"{prefix}_exp.npy")
            val_file = os.path.join(self.annotations_dir, f"{prefix}_val.npy")
            aro_file = os.path.join(self.annotations_dir, f"{prefix}_aro.npy")

            if not (os.path.exists(exp_file) and os.path.exists(val_file) and os.path.exists(aro_file)):
                continue

            exp = int(np.load(exp_file, allow_pickle=True).item())
            val = float(np.load(val_file, allow_pickle=True).item())
            aro = float(np.load(aro_file, allow_pickle=True).item())

            img_path = None
            for ext in [".jpg", ".png", ".jpeg"]:
                candidate = os.path.join(self.images_dir, f"{prefix}{ext}")
                if os.path.exists(candidate):
                    img_path = candidate
                    break
            if img_path is None:
                continue

            samples.append({
                "image": img_path,
                "expression": exp,
                "valence": val,
                "arousal": aro
            })

        return samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        img = Image.open(s["image"]).convert("RGB")
        img = self.transform(img)

        exp = torch.tensor(s["expression"], dtype=torch.long)
        val = torch.tensor(s["valence"], dtype=torch.float32)
        aro = torch.tensor(s["arousal"], dtype=torch.float32)

        return img, (exp, val, aro)
    
    def get_sample_for_viz(self, idx):
        """Get sample without normalization for visualization"""
        s = self.samples[idx]
        img = Image.open(s["image"]).convert("RGB")
        img_viz = self.viz_transform(img)
        
        return img_viz, s["expression"], s["valence"], s["arousal"], s["image"]


def get_dataloaders(images_dir, annotations_dir, img_size=(224,224),
                    batch_size=32, train_ratio=0.7, val_ratio=0.15):
    """
    Split dataset into train/val/test with ratios train_ratio:val_ratio:(1-train_ratio-val_ratio)
    """
    # Create datasets for all three splits
    train_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                                augment=True, split="train", split_ratio=train_ratio, seed=42)
    
    val_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                              augment=False, split="val", split_ratio=train_ratio, seed=42)
    
    test_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                               augment=False, split="test", split_ratio=train_ratio + val_ratio, seed=42)

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=0)
    val_loader   = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=0)
    test_loader  = DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=0)

    return train_loader, val_loader, test_loader, train_ds, val_ds, test_ds


# ============================================================
# Emotion-Specific Custom CNN - Enhanced for Emotion Detection
# ============================================================
class CustomCNN(nn.Module):
    """Emotion-specific CNN with attention mechanisms and specialized layers"""
    def __init__(self, num_classes=8, img_channels=3, dropout_rate=0.3):
        super().__init__()
        
        # Initial feature extraction with emotion-relevant filters
        self.initial_conv = nn.Sequential(
            nn.Conv2d(img_channels, 64, kernel_size=7, stride=2, padding=3),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
        )
        
        # Multi-scale feature extraction for facial expressions
        # Scale 1: Fine details (eyes, mouth corners)
        self.fine_features = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
        )
        
        # Scale 2: Medium details (facial regions)
        self.medium_features = nn.Sequential(
            nn.Conv2d(128, 256, kernel_size=5, padding=2),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
        )
        
        # Scale 3: Coarse details (overall face structure)
        self.coarse_features = nn.Sequential(
            nn.Conv2d(256, 512, kernel_size=7, padding=3),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
        )
        
        # Spatial attention mechanism for focusing on key facial regions
        self.spatial_attention = nn.Sequential(
            nn.Conv2d(512, 256, kernel_size=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 1, kernel_size=1),
            nn.Sigmoid()
        )
        
        # Channel attention mechanism
        self.channel_attention = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(512, 256, kernel_size=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 512, kernel_size=1),
            nn.Sigmoid()
        )
        
        # Emotion-specific feature enhancement
        self.emotion_enhancement = nn.Sequential(
            nn.Conv2d(512, 512, kernel_size=3, padding=1, groups=8),  # Grouped conv for efficiency
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 1024, kernel_size=1),  # Pointwise conv
            nn.BatchNorm2d(1024),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        
        # Pooling strategies for different tasks
        self.global_avg_pool = nn.AdaptiveAvgPool2d(1)
        self.global_max_pool = nn.AdaptiveMaxPool2d(1)
        
        # Multi-branch feature fusion
        feature_size = 1024 * 4 * 4  # From emotion_enhancement: 16384
        gap_gmp_size = 512 + 512     # GAP + GMP features: 1024
        total_features = feature_size + gap_gmp_size  # 16384 + 1024 = 17408
        
        self.feature_fusion = nn.Sequential(
            nn.Linear(total_features, 2048),
            nn.BatchNorm1d(2048),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(2048, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate * 0.5)
        )
        
        # Task-specific heads with emotion understanding
        # Expression head with hierarchical classification
        self.exp_head = nn.Sequential(
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(256, num_classes)
        )
        
        # Valence head (positive/negative emotion intensity)
        self.val_head = nn.Sequential(
            nn.Linear(1024, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(128, 1),
            nn.Tanh()  # Constrain output to [-1, 1]
        )
        
        # Arousal head (emotion activation level)
        self.aro_head = nn.Sequential(
            nn.Linear(1024, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(128, 1),
            nn.Tanh()  # Constrain output to [-1, 1]
        )
        
        # Initialize weights for emotion detection
        self._initialize_weights()
    
    def _initialize_weights(self):
        """Proper weight initialization for emotion detection"""
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                # He initialization for ReLU, Xavier for other activations
                if hasattr(m, 'groups') and m.groups > 1:
                    # Grouped convolutions
                    nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                else:
                    nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d) or isinstance(m, nn.BatchNorm1d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                # Special initialization for emotion tasks
                nn.init.xavier_uniform_(m.weight, gain=nn.init.calculate_gain('relu'))
                nn.init.constant_(m.bias, 0)
    
    def forward(self, x):
        # Initial feature extraction
        x = self.initial_conv(x)  # [B, 64, 56, 56]
        
        # Multi-scale feature extraction
        x = self.fine_features(x)    # [B, 128, 56, 56]
        x = F.max_pool2d(x, 2)       # [B, 128, 28, 28]
        
        x = self.medium_features(x)  # [B, 256, 28, 28]
        x = F.max_pool2d(x, 2)       # [B, 256, 14, 14]
        
        x = self.coarse_features(x)  # [B, 512, 14, 14]
        
        # Apply attention mechanisms
        spatial_att = self.spatial_attention(x)  # [B, 1, 14, 14]
        channel_att = self.channel_attention(x)  # [B, 512, 1, 1]
        
        # Apply attention
        x_attended = x * spatial_att * channel_att  # [B, 512, 14, 14]
        
        # Emotion-specific feature enhancement
        enhanced_features = self.emotion_enhancement(x_attended)  # [B, 1024, 4, 4]
        
        # Global pooling for additional features
        gap_features = self.global_avg_pool(x_attended).view(x.size(0), -1)  # [B, 512]
        gmp_features = self.global_max_pool(x_attended).view(x.size(0), -1)  # [B, 512]
        
        # Flatten enhanced features
        enhanced_flat = enhanced_features.view(enhanced_features.size(0), -1)  # [B, 1024*16]
        
        # Concatenate all features
        all_features = torch.cat([enhanced_flat, gap_features, gmp_features], dim=1)
        
        # Feature fusion
        fused_features = self.feature_fusion(all_features)  # [B, 1024]
        
        # Task-specific outputs
        exp_logits = self.exp_head(fused_features)
        val_out = self.val_head(fused_features).squeeze(1)
        aro_out = self.aro_head(fused_features).squeeze(1)
        
        return exp_logits, val_out, aro_out


# ============================================================
# Enhanced Transfer Learning Models
# ============================================================
class MultiTaskModel(nn.Module):
    """Enhanced Transfer learning models with regularization"""
    def __init__(self, backbone_name="resnet50", num_classes=8, pretrained=True, dropout_rate=0.3):
        super().__init__()

        if backbone_name == "resnet50":
            backbone = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1 if pretrained else None)
            in_features = backbone.fc.in_features
            backbone.fc = nn.Identity()
        elif backbone_name == "mobilenet_v2":
            backbone = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1 if pretrained else None)
            in_features = backbone.classifier[-1].in_features
            backbone.classifier = nn.Identity()
        else:
            raise ValueError("Unsupported backbone")

        self.backbone = backbone
        
        # Enhanced heads with regularization
        self.exp_head = nn.Sequential(
            nn.Linear(in_features, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(512, num_classes)
        )
        
        self.val_head = nn.Sequential(
            nn.Linear(in_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 1)
        )
        
        self.aro_head = nn.Sequential(
            nn.Linear(in_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 1)
        )

    def forward(self, x):
        feats = self.backbone(x)
        exp_logits = self.exp_head(feats)
        val_out = self.val_head(feats).squeeze(1)
        aro_out = self.aro_head(feats).squeeze(1)
        return exp_logits, val_out, aro_out


# ============================================================
# Early Stopping Implementation
# ============================================================
class EarlyStopping:
    """Early stopping to prevent overfitting"""
    def __init__(self, patience=7, min_delta=0.001, restore_best_weights=True):
        self.patience = patience
        self.min_delta = min_delta
        self.restore_best_weights = restore_best_weights
        self.best_loss = None
        self.counter = 0
        self.best_weights = None

    def __call__(self, val_loss, model):
        if self.best_loss is None:
            self.best_loss = val_loss
            self.save_checkpoint(model)
        elif val_loss < self.best_loss - self.min_delta:
            self.best_loss = val_loss
            self.counter = 0
            self.save_checkpoint(model)
        else:
            self.counter += 1

        if self.counter >= self.patience:
            if self.restore_best_weights:
                model.load_state_dict(self.best_weights)
            return True
        return False

    def save_checkpoint(self, model):
        """Save model checkpoint"""
        self.best_weights = model.state_dict().copy()


# ============================================================
# Model Summary Function
# ============================================================
def get_model_summary(model):
    """Get model parameter summary"""
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    
    # Estimate model size in MB (assuming float32)
    model_size_mb = total_params * 4 / (1024 * 1024)
    
    return {
        'total_params': total_params,
        'trainable_params': trainable_params,
        'model_size_mb': model_size_mb
    }


# ============================================================
# Fixed Advanced Evaluation Metrics
# ============================================================
def krippendorff_alpha(data, level_of_measurement='nominal'):
    """Calculate Krippendorff's Alpha for agreement - FIXED VERSION"""
    y_true, y_pred = data[0], data[1]
    
    if len(y_true) != len(y_pred):
        return 0.0
    
    n = len(y_true)
    if n < 2:
        return 0.0
    
    if level_of_measurement == 'nominal':
        # Count observed agreements
        observed_agreements = np.sum(y_true == y_pred)
        
        # Calculate expected agreements by chance
        unique_vals = np.unique(np.concatenate([y_true, y_pred]))
        expected_agreement = 0.0
        
        for val in unique_vals:
            # Marginal frequencies
            freq_true = np.sum(y_true == val) / n
            freq_pred = np.sum(y_pred == val) / n
            expected_agreement += freq_true * freq_pred
        
        # Calculate alpha
        observed_prop = observed_agreements / n
        
        if expected_agreement >= 1.0:
            return 1.0 if observed_prop == 1.0 else 0.0
        
        alpha = (observed_prop - expected_agreement) / (1 - expected_agreement)
        return max(0.0, min(1.0, alpha))  # Clamp between 0 and 1
    
    else:  # For ordinal/interval data
        # Simplified version for continuous data
        mae = np.mean(np.abs(y_true - y_pred))
        data_range = np.max(np.concatenate([y_true, y_pred])) - np.min(np.concatenate([y_true, y_pred]))
        
        if data_range == 0:
            return 1.0 if mae == 0 else 0.0
        
        # Normalize MAE by data range
        normalized_disagreement = mae / data_range
        alpha = 1.0 - 2 * normalized_disagreement  # Simple transformation
        return max(0.0, min(1.0, alpha))

def sign_agreement_metric(y_true, y_pred):
    """Sign Agreement Metric (SAGR) - penalizes incorrect sign alongside deviation"""
    same_sign = np.sign(y_true) == np.sign(y_pred)
    mae = np.mean(np.abs(y_true - y_pred))
    sign_penalty = np.mean(~same_sign) * 2.0
    sagr = 1.0 / (1.0 + mae + sign_penalty)
    return sagr

def concordance_correlation_coefficient(y_true, y_pred):
    """Concordance Correlation Coefficient (CCC)"""
    mean_true = np.mean(y_true)
    mean_pred = np.mean(y_pred)
    var_true = np.var(y_true)
    var_pred = np.var(y_pred)
    
    if len(y_true) > 1:
        r, _ = pearsonr(y_true, y_pred)
    else:
        r = 0
    
    numerator = 2 * r * np.sqrt(var_true) * np.sqrt(var_pred)
    denominator = var_true + var_pred + (mean_true - mean_pred)**2
    
    if denominator == 0:
        return 0
    
    ccc = numerator / denominator
    return ccc

def evaluate_model_comprehensive(model, data_loader, device, num_classes=8):
    """Comprehensive evaluation with all requested metrics"""
    model.eval()
    
    all_exp_preds = []
    all_exp_true = []
    all_exp_probs = []
    all_val_preds = []
    all_val_true = []
    all_aro_preds = []
    all_aro_true = []
    
    with torch.no_grad():
        for imgs, (exp, val, aro) in data_loader:
            imgs = imgs.to(device)
            exp_logits, val_pred, aro_pred = model(imgs)
            
            # Expression (categorical)
            exp_probs = torch.softmax(exp_logits, dim=1)
            _, exp_pred = torch.max(exp_logits, 1)
            
            all_exp_preds.extend(exp_pred.cpu().numpy())
            all_exp_true.extend(exp.numpy())
            all_exp_probs.extend(exp_probs.cpu().numpy())
            
            # Valence and Arousal (continuous)
            all_val_preds.extend(val_pred.cpu().numpy())
            all_val_true.extend(val.numpy())
            all_aro_preds.extend(aro_pred.cpu().numpy())
            all_aro_true.extend(aro.numpy())
    
    # Convert to numpy arrays
    all_exp_preds = np.array(all_exp_preds)
    all_exp_true = np.array(all_exp_true)
    all_exp_probs = np.array(all_exp_probs)
    all_val_preds = np.array(all_val_preds)
    all_val_true = np.array(all_val_true)
    all_aro_preds = np.array(all_aro_preds)
    all_aro_true = np.array(all_aro_true)
    
    # Categorical metrics (Expression)
    exp_metrics = {}
    exp_metrics['Accuracy'] = accuracy_score(all_exp_true, all_exp_preds)
    exp_metrics['F1-Score'] = f1_score(all_exp_true, all_exp_preds, average='macro')
    exp_metrics['Cohens Kappa'] = cohen_kappa_score(all_exp_true, all_exp_preds)
    exp_metrics['Krippendorfs Alpha'] = krippendorff_alpha([all_exp_true, all_exp_preds])
    
    try:
        exp_true_bin = label_binarize(all_exp_true, classes=range(num_classes))
        if exp_true_bin.shape[1] == 1:
            exp_metrics['AUC'] = roc_auc_score(all_exp_true, all_exp_probs[:, 1])
        else:
            exp_metrics['AUC'] = roc_auc_score(exp_true_bin, all_exp_probs, 
                                             multi_class='ovr', average='macro')
    except:
        exp_metrics['AUC'] = 0.0
    
    try:
        if exp_true_bin.shape[1] == 1:
            exp_metrics['AUC-PR'] = average_precision_score(all_exp_true, all_exp_probs[:, 1])
        else:
            exp_metrics['AUC-PR'] = average_precision_score(exp_true_bin, all_exp_probs, average='macro')
    except:
        exp_metrics['AUC-PR'] = 0.0
    
    # Continuous metrics (Valence)
    val_metrics = {}
    val_metrics['RMSE'] = np.sqrt(np.mean((all_val_true - all_val_preds)**2))
    if len(all_val_true) > 1:
        val_corr, _ = pearsonr(all_val_true, all_val_preds)
        val_metrics['CORR'] = val_corr if not np.isnan(val_corr) else 0.0
    else:
        val_metrics['CORR'] = 0.0
    val_metrics['SAGR'] = sign_agreement_metric(all_val_true, all_val_preds)
    val_metrics['CCC'] = concordance_correlation_coefficient(all_val_true, all_val_preds)
    
    # Continuous metrics (Arousal)
    aro_metrics = {}
    aro_metrics['RMSE'] = np.sqrt(np.mean((all_aro_true - all_aro_preds)**2))
    if len(all_aro_true) > 1:
        aro_corr, _ = pearsonr(all_aro_true, all_aro_preds)
        aro_metrics['CORR'] = aro_corr if not np.isnan(aro_corr) else 0.0
    else:
        aro_metrics['CORR'] = 0.0
    aro_metrics['SAGR'] = sign_agreement_metric(all_aro_true, all_aro_preds)
    aro_metrics['CCC'] = concordance_correlation_coefficient(all_aro_true, all_aro_preds)
    
    return {
        'expression': exp_metrics,
        'valence': val_metrics,
        'arousal': aro_metrics,
        'predictions': {
            'exp_true': all_exp_true,
            'exp_pred': all_exp_preds,
            'val_true': all_val_true,
            'val_pred': all_val_preds,
            'aro_true': all_aro_true,
            'aro_pred': all_aro_preds
        }
    }

def print_evaluation_results(results, model_name, dataset_type="Validation"):
    """Print formatted evaluation results"""
    print(f"\n{'='*60}")
    print(f"EVALUATION RESULTS FOR {model_name} - {dataset_type}")
    print(f"{'='*60}")
    
    # Expression (Categorical) Results
    print(f"\nExpression Classification Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*55}")
    exp_descriptions = {
        'Accuracy': 'Accuracy',
        'F1-Score': 'F1-Score',
        'Cohens Kappa': 'Statistical measure of agreement',
        'Krippendorfs Alpha': 'Statistical measure of agreement',
        'AUC': 'Area under ROC curve',
        'AUC-PR': 'Area under Precision-Recall curve'
    }
    
    for metric, value in results['expression'].items():
        desc = exp_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")
    
    # Valence (Continuous) Results
    print(f"\nValence Regression Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*70}")
    val_descriptions = {
        'RMSE': 'Root Mean Square Error',
        'CORR': 'Pearson Correlation',
        'SAGR': 'Sign Agreement Metric',
        'CCC': 'Concordance Correlation Coefficient'
    }
    
    for metric, value in results['valence'].items():
        desc = val_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")
    
    # Arousal (Continuous) Results
    print(f"\nArousal Regression Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*70}")
    
    for metric, value in results['arousal'].items():
        desc = val_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")


# ============================================================
# Sample Visualization Function
# ============================================================
def visualize_sample_predictions(model, dataset, device, num_samples=8, 
                               expression_names=None):
    """Visualize sample predictions with ground truth"""
    if expression_names is None:
        expression_names = ['Neutral', 'Happiness', 'Sadness', 'Surprise', 
                          'Fear', 'Disgust', 'Anger', 'Contempt']
    
    model.eval()
    
    # Select random samples
    indices = random.sample(range(len(dataset)), min(num_samples, len(dataset)))
    
    fig, axes = plt.subplots(2, 4, figsize=(20, 10))
    axes = axes.ravel()
    
    with torch.no_grad():
        for i, idx in enumerate(indices):
            # Get sample for visualization
            img_viz, true_exp, true_val, true_aro, img_path = dataset.get_sample_for_viz(idx)
            
            # Get model prediction
            img_tensor, (exp_tensor, val_tensor, aro_tensor) = dataset[idx]
            img_batch = img_tensor.unsqueeze(0).to(device)
            
            exp_logits, val_pred, aro_pred = model(img_batch)
            exp_probs = torch.softmax(exp_logits, dim=1)
            _, exp_pred = torch.max(exp_logits, 1)
            
            exp_pred = exp_pred.cpu().item()
            val_pred = val_pred.cpu().item()
            aro_pred = aro_pred.cpu().item()
            exp_conf = exp_probs[0][exp_pred].cpu().item()
            
            # Display image
            img_np = img_viz.permute(1, 2, 0).numpy()
            axes[i].imshow(img_np)
            axes[i].axis('off')
            
            # Create title with predictions
            title = f"True: {expression_names[true_exp]}\n"
            title += f"Pred: {expression_names[exp_pred]} ({exp_conf:.2f})\n"
            title += f"Val: {true_val:.2f}→{val_pred:.2f}\n"
            title += f"Aro: {true_aro:.2f}→{aro_pred:.2f}"
            
            # Color code: green for correct, red for incorrect
            color = 'green' if exp_pred == true_exp else 'red'
            axes[i].set_title(title, fontsize=10, color=color)
    
    plt.tight_layout()
    plt.suptitle("Sample Predictions vs Ground Truth", fontsize=16, y=1.02)
    plt.show()


# ============================================================
# Enhanced Training with Better Learning Rate Schedule
# ============================================================
def train_model(model, train_loader, val_loader, device, epochs=50, lr=0.001, 
                model_name="Model", patience=15, weight_decay=1e-4):
    """Enhanced training with better learning dynamics for custom CNN"""
    ce_loss = nn.CrossEntropyLoss(label_smoothing=0.1)  # Label smoothing for regularization
    mse_loss = nn.MSELoss()

    # Optimizer selection based on model type
    if "Custom" in model_name:
        # Custom CNN benefits from Adam with specific parameters
        optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay, 
                              betas=(0.9, 0.999), eps=1e-8)
        # More aggressive learning rate schedule for custom CNN
        scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.7, patience=3, 
                                    min_lr=1e-6, verbose=True)
    else:
        # Transfer learning models use AdamW
        optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
        scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, verbose=True)
    
    # Early stopping with different patience for custom CNN
    early_stopping = EarlyStopping(patience=patience, min_delta=0.001)

    history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
    
    print(f"\nTraining {model_name}...")
    print(f"Initial Learning Rate: {lr}")
    print(f"Weight Decay: {weight_decay}")
    print(f"Early Stopping Patience: {patience}")
    
    best_val_acc = 0.0

    for epoch in range(epochs):
        # Training phase
        model.train()
        total_loss, correct, total = 0, 0, 0

        for imgs, (exp, val, aro) in train_loader:
            imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)

            optimizer.zero_grad()
            exp_logits, val_pred, aro_pred = model(imgs)

            # Multi-task loss with balanced weighting
            loss_exp = ce_loss(exp_logits, exp)
            loss_val = mse_loss(val_pred, val)
            loss_aro = mse_loss(aro_pred, aro)
            
            # Balanced loss weighting (expression is primary task)
            if "Custom" in model_name:
                # Custom CNN focuses more on expression classification
                loss = 3.0 * loss_exp + 0.5 * loss_val + 0.5 * loss_aro
            else:
                loss = 2.0 * loss_exp + loss_val + loss_aro

            loss.backward()
            
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            
            optimizer.step()

            total_loss += loss.item() * imgs.size(0)
            _, preds = torch.max(exp_logits, 1)
            correct += (preds == exp).sum().item()
            total += exp.size(0)

        train_loss = total_loss / total
        train_acc = correct / total

        # Validation phase
        model.eval()
        val_loss, val_correct, val_total = 0, 0, 0
        with torch.no_grad():
            for imgs, (exp, val, aro) in val_loader:
                imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)
                exp_logits, val_pred, aro_pred = model(imgs)

                loss_exp = ce_loss(exp_logits, exp)
                loss_val = mse_loss(val_pred, val)
                loss_aro = mse_loss(aro_pred, aro)
                
                if "Custom" in model_name:
                    loss = 3.0 * loss_exp + 0.5 * loss_val + 0.5 * loss_aro
                else:
                    loss = 2.0 * loss_exp + loss_val + loss_aro

                val_loss += loss.item() * imgs.size(0)
                _, preds = torch.max(exp_logits, 1)
                val_correct += (preds == exp).sum().item()
                val_total += exp.size(0)

        val_loss /= val_total
        val_acc = val_correct / val_total
        
        # Learning rate scheduling
        scheduler.step(val_loss)

        # Track best validation accuracy
        if val_acc > best_val_acc:
            best_val_acc = val_acc

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)

        # Print progress every 3 epochs or last epoch
        if epoch % 3 == 0 or epoch == epochs - 1:
            print(f"Epoch {epoch+1}/{epochs} | "
                  f"Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | "
                  f"Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.6f}")
        
        # Early stopping check
        if early_stopping(val_loss, model):
            print(f"Early stopping triggered at epoch {epoch+1}")
            break
    
    print(f"{model_name} training completed! Best Val Acc: {best_val_acc:.4f}")
    return history


def plot_history(history, title="Training Curves"):
    """Enhanced plotting function"""
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.plot(history["train_loss"], label="Train Loss", linewidth=2)
    plt.plot(history["val_loss"], label="Val Loss", linewidth=2)
    plt.legend()
    plt.title("Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 3, 2)
    plt.plot(history["train_acc"], label="Train Acc", linewidth=2)
    plt.plot(history["val_acc"], label="Val Acc", linewidth=2)
    plt.legend()
    plt.title("Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 3, 3)
    # Overfitting indicator
    overfitting = np.array(history["train_acc"]) - np.array(history["val_acc"])
    plt.plot(overfitting, label="Overfitting Gap", color='red', linewidth=2)
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.legend()
    plt.title("Overfitting Monitor")
    plt.xlabel("Epoch")
    plt.ylabel("Train Acc - Val Acc")
    plt.grid(True, alpha=0.3)
    
    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()


# ============================================================
# Debugging and Analysis Functions
# ============================================================
def analyze_model_gradients(model, data_loader, device):
    """Analyze gradient flow in the model"""
    model.train()
    grad_norms = []
    
    for imgs, (exp, val, aro) in data_loader:
        imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)
        
        model.zero_grad()
        exp_logits, val_pred, aro_pred = model(imgs)
        
        loss_exp = F.cross_entropy(exp_logits, exp)
        loss_exp.backward()
        
        total_norm = 0
        for p in model.parameters():
            if p.grad is not None:
                param_norm = p.grad.data.norm(2)
                total_norm += param_norm.item() ** 2
        total_norm = total_norm ** (1. / 2)
        grad_norms.append(total_norm)
        
        break  # Just check first batch
    
    return grad_norms[0] if grad_norms else 0


def check_model_outputs(model, data_loader, device):
    """Check model output distributions"""
    model.eval()
    exp_outputs = []
    val_outputs = []
    aro_outputs = []
    
    with torch.no_grad():
        for imgs, (exp, val, aro) in data_loader:
            imgs = imgs.to(device)
            exp_logits, val_pred, aro_pred = model(imgs)
            
            exp_outputs.extend(torch.softmax(exp_logits, dim=1).cpu().numpy())
            val_outputs.extend(val_pred.cpu().numpy())
            aro_outputs.extend(aro_pred.cpu().numpy())
            
            break  # Just check first batch
    
    exp_outputs = np.array(exp_outputs)
    val_outputs = np.array(val_outputs)
    aro_outputs = np.array(aro_outputs)
    
    print(f"Expression logits - Mean: {exp_outputs.mean():.4f}, Std: {exp_outputs.std():.4f}")
    print(f"Valence outputs - Mean: {val_outputs.mean():.4f}, Std: {val_outputs.std():.4f}")
    print(f"Arousal outputs - Mean: {aro_outputs.mean():.4f}, Std: {aro_outputs.std():.4f}")


# ============================================================
# Main Training Function with All Models
# ============================================================
def main():
    # Configuration - UPDATE THESE PATHS FOR YOUR LOCAL SETUP
    IMAGES_DIR = "./Dataset/images"  # Update this path
    ANNOTATIONS_DIR = "./Dataset/annotations"  # Update this path
    IMG_SIZE = (224, 224)
    BATCH_SIZE = 32
    EPOCHS = 10 
    
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")
    
    # Load data with train/val/test split
    train_loader, val_loader, test_loader, train_ds, val_ds, test_ds = get_dataloaders(
        IMAGES_DIR, ANNOTATIONS_DIR, img_size=IMG_SIZE, batch_size=BATCH_SIZE)
    
    print(f"Dataset loaded:")
    print(f"   • Train: {len(train_ds)} samples")
    print(f"   • Validation: {len(val_ds)} samples") 
    print(f"   • Test: {len(test_ds)} samples")
    
    # Expression class names (adjust according to your dataset)
    expression_names = ['Neutral', 'Happiness', 'Sadness', 'Surprise', 
                       'Fear', 'Disgust', 'Anger', 'Contempt']
    
    # Store all results
    all_results = {}
    all_models = {}
    
    # ========================
    # BASELINE 1: ResNet50
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING RESNET50 MODEL")
    print(f"{'='*70}")
    
    model_resnet = MultiTaskModel("resnet50", num_classes=8, dropout_rate=0.3).to(device)
    resnet_summary = get_model_summary(model_resnet)
    print(f"ResNet50 Model Summary:")
    print(f"   • Total Parameters: {resnet_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {resnet_summary['trainable_params']:,}")
    print(f"   • Model Size: {resnet_summary['model_size_mb']:.1f} MB")
    
    hist_resnet = train_model(model_resnet, train_loader, val_loader, device, 
                             epochs=EPOCHS, model_name="ResNet50", patience=10,
                             lr=0.0001, weight_decay=1e-4)
    plot_history(hist_resnet, title="ResNet50 Training Curves")
    
    # Validation evaluation
    results_resnet_val = evaluate_model_comprehensive(model_resnet, val_loader, device)
    print_evaluation_results(results_resnet_val, "RESNET50", "Validation")
    
    # Test evaluation
    results_resnet_test = evaluate_model_comprehensive(model_resnet, test_loader, device)
    print_evaluation_results(results_resnet_test, "RESNET50", "Test")
    
    # Store results
    all_results['ResNet50'] = {
        'validation': results_resnet_val,
        'test': results_resnet_test,
        'history': hist_resnet,
        'model_summary': resnet_summary
    }
    all_models['ResNet50'] = model_resnet
    
    # ========================
    # BASELINE 2: MobileNetV2
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING MOBILENETV2 MODEL")
    print(f"{'='*70}")
    
    model_mobilenet = MultiTaskModel("mobilenet_v2", num_classes=8, dropout_rate=0.3).to(device)
    mobilenet_summary = get_model_summary(model_mobilenet)
    print(f"MobileNetV2 Model Summary:")
    print(f"   • Total Parameters: {mobilenet_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {mobilenet_summary['trainable_params']:,}")
    print(f"   • Model Size: {mobilenet_summary['model_size_mb']:.1f} MB")
    
    hist_mobilenet = train_model(model_mobilenet, train_loader, val_loader, device, 
                                epochs=EPOCHS, model_name="MobileNetV2", patience=10,
                                lr=0.0001, weight_decay=1e-4)
    plot_history(hist_mobilenet, title="MobileNetV2 Training Curves")
    
    # Validation evaluation
    results_mobilenet_val = evaluate_model_comprehensive(model_mobilenet, val_loader, device)
    print_evaluation_results(results_mobilenet_val, "MOBILENETV2", "Validation")
    
    # Test evaluation
    results_mobilenet_test = evaluate_model_comprehensive(model_mobilenet, test_loader, device)
    print_evaluation_results(results_mobilenet_test, "MOBILENETV2", "Test")
    
    # Store results
    all_results['MobileNetV2'] = {
        'validation': results_mobilenet_val,
        'test': results_mobilenet_test,
        'history': hist_mobilenet,
        'model_summary': mobilenet_summary
    }
    all_models['MobileNetV2'] = model_mobilenet
    
    # ========================
    # IMPROVED CUSTOM CNN MODEL
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING IMPROVED CUSTOM CNN MODEL")
    print(f"{'='*70}")
    
    model_custom = CustomCNN(num_classes=8, dropout_rate=0.3).to(device)
    custom_summary = get_model_summary(model_custom)
    print(f"Custom CNN Model Summary:")
    print(f"   • Total Parameters: {custom_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {custom_summary['trainable_params']:,}")
    print(f"   • Model Size: {custom_summary['model_size_mb']:.1f} MB")
    
    # Debug model before training
    print(f"\nDEBUG: Checking model outputs before training...")
    check_model_outputs(model_custom, train_loader, device)
    grad_norm = analyze_model_gradients(model_custom, train_loader, device)
    print(f"Initial gradient norm: {grad_norm:.4f}")
    
    # Custom CNN training with optimized parameters
    hist_custom = train_model(model_custom, train_loader, val_loader, device, 
                             epochs=EPOCHS, lr=0.0001, model_name="Custom CNN", 
                             patience=15, weight_decay=1e-4)
    plot_history(hist_custom, title="Custom CNN Training Curves")
    
    # Validation evaluation
    results_custom_val = evaluate_model_comprehensive(model_custom, val_loader, device)
    print_evaluation_results(results_custom_val, "CUSTOM CNN", "Validation")
    
    # Test evaluation
    results_custom_test = evaluate_model_comprehensive(model_custom, test_loader, device)
    print_evaluation_results(results_custom_test, "CUSTOM CNN", "Test")
    
    # Store results
    all_results['Custom CNN'] = {
        'validation': results_custom_val,
        'test': results_custom_test,
        'history': hist_custom,
        'model_summary': custom_summary
    }
    all_models['Custom CNN'] = model_custom
    
    # Visualize sample predictions for each model
    for model_name, model in all_models.items():
        print(f"\nSAMPLE PREDICTIONS - {model_name.upper()}:")
        visualize_sample_predictions(model, test_ds, device, 
                                    expression_names=expression_names)
    
    # ========================
    # COMPREHENSIVE MODEL COMPARISON
    # ========================
    print(f"\n{'='*80}")
    print("COMPREHENSIVE MODEL COMPARISON")
    print(f"{'='*80}")
    
    # Create comparison tables
    models = ['ResNet50', 'MobileNetV2', 'Custom CNN']
    
    # Model Architecture Comparison
    print(f"\nMODEL ARCHITECTURE COMPARISON:")
    print(f"{'Model':<15} {'Parameters':<15} {'Size (MB)':<12} {'Type'}")
    print(f"{'-'*65}")
    for model_name in models:
        summary = all_results[model_name]['model_summary']
        model_type = "Transfer Learning" if model_name != "Custom CNN" else "Custom CNN"
        print(f"{model_name:<15} {summary['total_params']:>14,} {summary['model_size_mb']:>11.1f} {model_type}")
    
    # Validation Results Comparison
    print(f"\nVALIDATION RESULTS COMPARISON:")
    val_comparison_df = pd.DataFrame({
        model: [
            all_results[model]['validation']['expression']['Accuracy'],
            all_results[model]['validation']['expression']['F1-Score'],
            all_results[model]['validation']['expression']['AUC'],
            all_results[model]['validation']['valence']['RMSE'],
            all_results[model]['validation']['valence']['CORR'],
            all_results[model]['validation']['valence']['CCC'],
            all_results[model]['validation']['arousal']['RMSE'],
            all_results[model]['validation']['arousal']['CORR'],
            all_results[model]['validation']['arousal']['CCC']
        ] for model in models
    }, index=[
        'Exp_Accuracy', 'Exp_F1', 'Exp_AUC',
        'Val_RMSE', 'Val_CORR', 'Val_CCC',
        'Aro_RMSE', 'Aro_CORR', 'Aro_CCC'
    ])
    
    print(val_comparison_df.round(4))
    
    # Test Results Comparison
    print(f"\nTEST RESULTS COMPARISON:")
    test_comparison_df = pd.DataFrame({
        model: [
            all_results[model]['test']['expression']['Accuracy'],
            all_results[model]['test']['expression']['F1-Score'],
            all_results[model]['test']['expression']['AUC'],
            all_results[model]['test']['valence']['RMSE'],
            all_results[model]['test']['valence']['CORR'],
            all_results[model]['test']['valence']['CCC'],
            all_results[model]['test']['arousal']['RMSE'],
            all_results[model]['test']['arousal']['CORR'],
            all_results[model]['test']['arousal']['CCC']
        ] for model in models
    }, index=[
        'Exp_Accuracy', 'Exp_F1', 'Exp_AUC',
        'Val_RMSE', 'Val_CORR', 'Val_CCC',
        'Aro_RMSE', 'Aro_CORR', 'Aro_CCC'
    ])
    
    print(test_comparison_df.round(4))
    
    # Best model identification
    test_accuracies = {model: all_results[model]['test']['expression']['Accuracy'] 
                      for model in models}
    best_model = max(test_accuracies, key=test_accuracies.get)
    
    print(f"\nBEST MODEL: {best_model}")
    print(f"   • Test Accuracy: {test_accuracies[best_model]:.4f}")
    print(f"   • Model Size: {all_results[best_model]['model_summary']['model_size_mb']:.1f} MB")
    print(f"   • Parameters: {all_results[best_model]['model_summary']['total_params']:,}")
    
    # Training curves comparison
    plt.figure(figsize=(18, 12))
    
    # Plot training losses
    plt.subplot(3, 2, 1)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['train_loss'], label=f'{model_name} Train', linewidth=2)
    plt.title('Training Loss Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot validation losses
    plt.subplot(3, 2, 2)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['val_loss'], label=f'{model_name} Val', linewidth=2)
    plt.title('Validation Loss Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot training accuracies
    plt.subplot(3, 2, 3)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['train_acc'], label=f'{model_name} Train', linewidth=2)
    plt.title('Training Accuracy Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot validation accuracies
    plt.subplot(3, 2, 4)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['val_acc'], label=f'{model_name} Val', linewidth=2)
    plt.title('Validation Accuracy Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot overfitting comparison
    plt.subplot(3, 2, 5)
    for model_name in models:
        hist = all_results[model_name]['history']
        overfitting = np.array(hist['train_acc']) - np.array(hist['val_acc'])
        plt.plot(overfitting, label=f'{model_name}', linewidth=2)
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.title('Overfitting Comparison (Train - Val Accuracy)')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy Difference')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Performance vs efficiency scatter
    plt.subplot(3, 2, 6)
    for model_name in models:
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        size_mb = all_results[model_name]['model_summary']['model_size_mb']
        plt.scatter(size_mb, test_acc, s=200, label=model_name, alpha=0.7)
        plt.text(size_mb, test_acc, model_name, fontsize=9, ha='center', va='bottom')
    plt.title('Performance vs Model Size')
    plt.xlabel('Model Size (MB)')
    plt.ylabel('Test Accuracy')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.suptitle('Comprehensive Model Analysis', fontsize=16, y=0.98)
    plt.show()
    
    # Performance vs Efficiency Analysis
    print(f"\nPERFORMANCE VS EFFICIENCY ANALYSIS:")
    print(f"{'Model':<15} {'Test Acc':<10} {'Size (MB)':<12} {'Params (M)':<12} {'Efficiency'}")
    print(f"{'-'*75}")
    for model_name in models:
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        size_mb = all_results[model_name]['model_summary']['model_size_mb']
        params_m = all_results[model_name]['model_summary']['total_params'] / 1e6
        efficiency = test_acc / size_mb  # Accuracy per MB
        print(f"{model_name:<15} {test_acc:<10.4f} {size_mb:<12.1f} {params_m:<12.1f} {efficiency:<10.4f}")
    
    # Overfitting Analysis
    print(f"\nOVERFITTING ANALYSIS:")
    print(f"{'Model':<15} {'Val Acc':<10} {'Test Acc':<10} {'Gap':<10} {'Status'}")
    print(f"{'-'*60}")
    for model_name in models:
        val_acc = all_results[model_name]['validation']['expression']['Accuracy']
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        gap = val_acc - test_acc
        status = "Overfit" if gap > 0.05 else "Good" if gap > -0.02 else "Underfit"
        print(f"{model_name:<15} {val_acc:<10.4f} {test_acc:<10.4f} {gap:<10.4f} {status}")
    
    # Final summary and recommendations
    print(f"\nKEY INSIGHTS:")
    
    # Find most efficient model
    efficiencies = {model: all_results[model]['test']['expression']['Accuracy'] / 
                   all_results[model]['model_summary']['model_size_mb'] 
                   for model in models}
    most_efficient = max(efficiencies, key=efficiencies.get)
    smallest_model = min(models, key=lambda x: all_results[x]['model_summary']['model_size_mb'])
    
    print(f"   • Best Overall Performance: {best_model}")
    print(f"   • Most Efficient Model: {most_efficient}")
    print(f"   • Smallest Model: {smallest_model}")
    
    # Custom CNN specific analysis
    custom_acc = all_results['Custom CNN']['test']['expression']['Accuracy']
    if custom_acc > 0.25:  # Better than random + some margin
        print(f"   • Custom CNN Performance: GOOD ({custom_acc:.4f})")
    else:
        print(f"   • Custom CNN Performance: NEEDS IMPROVEMENT ({custom_acc:.4f})")
    
    # Regularization effectiveness
    print(f"\nREGULARIZATION EFFECTIVENESS:")
    for model_name in models:
        hist = all_results[model_name]['history']
        final_gap = hist['train_acc'][-1] - hist['val_acc'][-1]
        print(f"   • {model_name}: Final overfitting gap = {final_gap:.4f}")
    
    return all_models, all_results


if __name__ == "__main__":
    models, results = main()
```

```python
# ============================================================
# Enhanced Multi-Task Emotion Recognition with Fixed Custom CNN
# ============================================================

import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import (accuracy_score, f1_score, cohen_kappa_score, 
                           roc_auc_score, average_precision_score, 
                           confusion_matrix, classification_report)
from sklearn.preprocessing import label_binarize
import pandas as pd
from scipy.stats import pearsonr

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
import torchvision.models as models
from torch.optim.lr_scheduler import ReduceLROnPlateau
import torch.nn.functional as F


# ============================================================
# Dataset Loader (Enhanced with better augmentation)
# ============================================================
class MultiTaskDataset(Dataset):
    def __init__(self, images_dir: str, annotations_dir: str, img_size=(224,224),
                 augment=False, split="train", split_ratio=0.8, seed=42):
        self.images_dir = images_dir
        self.annotations_dir = annotations_dir
        self.img_size = img_size
        self.augment = augment

        all_samples = self._gather_samples()
        random.seed(seed)
        random.shuffle(all_samples)
        
        # Handle three-way split: train/val/test
        if split == "train":
            split_idx = int(len(all_samples) * split_ratio)
            self.samples = all_samples[:split_idx]
        elif split == "val":
            train_split = int(len(all_samples) * 0.7)  # 70% train
            val_split = int(len(all_samples) * 0.85)   # 15% val (70% + 15%)
            self.samples = all_samples[train_split:val_split]
        else:  # test
            test_split = int(len(all_samples) * 0.85)  # 15% test (remaining)
            self.samples = all_samples[test_split:]

        # Enhanced transforms with better augmentation
        base_tfms = [
            T.Resize(self.img_size),
            T.ToTensor(),
            T.Normalize(mean=[0.485, 0.456, 0.406],
                        std=[0.229, 0.224, 0.225])
        ]

        if self.augment and split == "train":
            self.transform = T.Compose([
                T.Resize((int(self.img_size[0] * 1.1), int(self.img_size[1] * 1.1))),
                T.RandomCrop(self.img_size),
                T.RandomHorizontalFlip(p=0.5),
                T.RandomRotation(15),
                T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05),
                T.RandomGrayscale(p=0.05),
                T.ToTensor(),
                T.Normalize(mean=[0.485, 0.456, 0.406],
                            std=[0.229, 0.224, 0.225])
            ])
        else:
            self.transform = T.Compose(base_tfms)
            
        # Store original transform for visualization
        self.viz_transform = T.Compose([
            T.Resize(self.img_size),
            T.ToTensor(),
        ])

    def _gather_samples(self):
        samples = []
        files = os.listdir(self.annotations_dir)
        prefixes = sorted(set(f.split("_")[0] for f in files))

        for prefix in prefixes:
            exp_file = os.path.join(self.annotations_dir, f"{prefix}_exp.npy")
            val_file = os.path.join(self.annotations_dir, f"{prefix}_val.npy")
            aro_file = os.path.join(self.annotations_dir, f"{prefix}_aro.npy")

            if not (os.path.exists(exp_file) and os.path.exists(val_file) and os.path.exists(aro_file)):
                continue

            exp = int(np.load(exp_file, allow_pickle=True).item())
            val = float(np.load(val_file, allow_pickle=True).item())
            aro = float(np.load(aro_file, allow_pickle=True).item())

            img_path = None
            for ext in [".jpg", ".png", ".jpeg"]:
                candidate = os.path.join(self.images_dir, f"{prefix}{ext}")
                if os.path.exists(candidate):
                    img_path = candidate
                    break
            if img_path is None:
                continue

            samples.append({
                "image": img_path,
                "expression": exp,
                "valence": val,
                "arousal": aro
            })

        return samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        img = Image.open(s["image"]).convert("RGB")
        img = self.transform(img)

        exp = torch.tensor(s["expression"], dtype=torch.long)
        val = torch.tensor(s["valence"], dtype=torch.float32)
        aro = torch.tensor(s["arousal"], dtype=torch.float32)

        return img, (exp, val, aro)
    
    def get_sample_for_viz(self, idx):
        """Get sample without normalization for visualization"""
        s = self.samples[idx]
        img = Image.open(s["image"]).convert("RGB")
        img_viz = self.viz_transform(img)
        
        return img_viz, s["expression"], s["valence"], s["arousal"], s["image"]


def get_dataloaders(images_dir, annotations_dir, img_size=(224,224),
                    batch_size=32, train_ratio=0.7, val_ratio=0.15):
    """
    Split dataset into train/val/test with ratios train_ratio:val_ratio:(1-train_ratio-val_ratio)
    """
    # Create datasets for all three splits
    train_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                                augment=True, split="train", split_ratio=train_ratio, seed=42)
    
    val_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                              augment=False, split="val", split_ratio=train_ratio, seed=42)
    
    test_ds = MultiTaskDataset(images_dir, annotations_dir, img_size=img_size,
                               augment=False, split="test", split_ratio=train_ratio + val_ratio, seed=42)

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=2)
    val_loader   = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=2)
    test_loader  = DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=2)

    return train_loader, val_loader, test_loader, train_ds, val_ds, test_ds


# ============================================================
# Emotion-Specific Custom CNN - Enhanced for Emotion Detection
# ============================================================
class CustomCNN(nn.Module):
    """Emotion-specific CNN with attention mechanisms and specialized layers"""
    def __init__(self, num_classes=8, img_channels=3, dropout_rate=0.3):
        super().__init__()
        
        # Initial feature extraction with emotion-relevant filters
        self.initial_conv = nn.Sequential(
            nn.Conv2d(img_channels, 64, kernel_size=7, stride=2, padding=3),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
        )
        
        # Multi-scale feature extraction for facial expressions
        # Scale 1: Fine details (eyes, mouth corners)
        self.fine_features = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
        )
        
        # Scale 2: Medium details (facial regions)
        self.medium_features = nn.Sequential(
            nn.Conv2d(128, 256, kernel_size=5, padding=2),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
        )
        
        # Scale 3: Coarse details (overall face structure)
        self.coarse_features = nn.Sequential(
            nn.Conv2d(256, 512, kernel_size=7, padding=3),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
        )
        
        # Spatial attention mechanism for focusing on key facial regions
        self.spatial_attention = nn.Sequential(
            nn.Conv2d(512, 256, kernel_size=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 1, kernel_size=1),
            nn.Sigmoid()
        )
        
        # Channel attention mechanism
        self.channel_attention = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(512, 256, kernel_size=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 512, kernel_size=1),
            nn.Sigmoid()
        )
        
        # Emotion-specific feature enhancement
        self.emotion_enhancement = nn.Sequential(
            nn.Conv2d(512, 512, kernel_size=3, padding=1, groups=8),  # Grouped conv for efficiency
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 1024, kernel_size=1),  # Pointwise conv
            nn.BatchNorm2d(1024),
            nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        
        # Pooling strategies for different tasks
        self.global_avg_pool = nn.AdaptiveAvgPool2d(1)
        self.global_max_pool = nn.AdaptiveMaxPool2d(1)
        
        # Multi-branch feature fusion
        feature_size = 1024 * 4 * 4  # From emotion_enhancement: 16384
        gap_gmp_size = 512 + 512     # GAP + GMP features: 1024
        total_features = feature_size + gap_gmp_size  # 16384 + 1024 = 17408
        
        self.feature_fusion = nn.Sequential(
            nn.Linear(total_features, 2048),
            nn.BatchNorm1d(2048),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(2048, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate * 0.5)
        )
        
        # Task-specific heads with emotion understanding
        # Expression head with hierarchical classification
        self.exp_head = nn.Sequential(
            nn.Linear(1024, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(256, num_classes)
        )
        
        # Valence head (positive/negative emotion intensity)
        self.val_head = nn.Sequential(
            nn.Linear(1024, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(128, 1),
            nn.Tanh()  # Constrain output to [-1, 1]
        )
        
        # Arousal head (emotion activation level)
        self.aro_head = nn.Sequential(
            nn.Linear(1024, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(128, 1),
            nn.Tanh()  # Constrain output to [-1, 1]
        )
        
        # Initialize weights for emotion detection
        self._initialize_weights()
    
    def _initialize_weights(self):
        """Proper weight initialization for emotion detection"""
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                # He initialization for ReLU, Xavier for other activations
                if hasattr(m, 'groups') and m.groups > 1:
                    # Grouped convolutions
                    nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                else:
                    nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d) or isinstance(m, nn.BatchNorm1d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                # Special initialization for emotion tasks
                nn.init.xavier_uniform_(m.weight, gain=nn.init.calculate_gain('relu'))
                nn.init.constant_(m.bias, 0)
    
    def forward(self, x):
        # Initial feature extraction
        x = self.initial_conv(x)  # [B, 64, 56, 56]
        
        # Multi-scale feature extraction
        x = self.fine_features(x)    # [B, 128, 56, 56]
        x = F.max_pool2d(x, 2)       # [B, 128, 28, 28]
        
        x = self.medium_features(x)  # [B, 256, 28, 28]
        x = F.max_pool2d(x, 2)       # [B, 256, 14, 14]
        
        x = self.coarse_features(x)  # [B, 512, 14, 14]
        
        # Apply attention mechanisms
        spatial_att = self.spatial_attention(x)  # [B, 1, 14, 14]
        channel_att = self.channel_attention(x)  # [B, 512, 1, 1]
        
        # Apply attention
        x_attended = x * spatial_att * channel_att  # [B, 512, 14, 14]
        
        # Emotion-specific feature enhancement
        enhanced_features = self.emotion_enhancement(x_attended)  # [B, 1024, 4, 4]
        
        # Global pooling for additional features
        gap_features = self.global_avg_pool(x_attended).view(x.size(0), -1)  # [B, 512]
        gmp_features = self.global_max_pool(x_attended).view(x.size(0), -1)  # [B, 512]
        
        # Flatten enhanced features
        enhanced_flat = enhanced_features.view(enhanced_features.size(0), -1)  # [B, 1024*16]
        
        # Concatenate all features
        all_features = torch.cat([enhanced_flat, gap_features, gmp_features], dim=1)
        
        # Feature fusion
        fused_features = self.feature_fusion(all_features)  # [B, 1024]
        
        # Task-specific outputs
        exp_logits = self.exp_head(fused_features)
        val_out = self.val_head(fused_features).squeeze(1)
        aro_out = self.aro_head(fused_features).squeeze(1)
        
        return exp_logits, val_out, aro_out


# ============================================================
# Enhanced Transfer Learning Models
# ============================================================
class MultiTaskModel(nn.Module):
    """Enhanced Transfer learning models with regularization"""
    def __init__(self, backbone_name="resnet50", num_classes=8, pretrained=True, dropout_rate=0.3):
        super().__init__()

        if backbone_name == "resnet50":
            backbone = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1 if pretrained else None)
            in_features = backbone.fc.in_features
            backbone.fc = nn.Identity()
        elif backbone_name == "mobilenet_v2":
            backbone = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1 if pretrained else None)
            in_features = backbone.classifier[-1].in_features
            backbone.classifier = nn.Identity()
        else:
            raise ValueError("Unsupported backbone")

        self.backbone = backbone
        
        # Enhanced heads with regularization
        self.exp_head = nn.Sequential(
            nn.Linear(in_features, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(512, num_classes)
        )
        
        self.val_head = nn.Sequential(
            nn.Linear(in_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 1)
        )
        
        self.aro_head = nn.Sequential(
            nn.Linear(in_features, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 1)
        )

    def forward(self, x):
        feats = self.backbone(x)
        exp_logits = self.exp_head(feats)
        val_out = self.val_head(feats).squeeze(1)
        aro_out = self.aro_head(feats).squeeze(1)
        return exp_logits, val_out, aro_out


# ============================================================
# Early Stopping Implementation
# ============================================================
class EarlyStopping:
    """Early stopping to prevent overfitting"""
    def __init__(self, patience=7, min_delta=0.001, restore_best_weights=True):
        self.patience = patience
        self.min_delta = min_delta
        self.restore_best_weights = restore_best_weights
        self.best_loss = None
        self.counter = 0
        self.best_weights = None

    def __call__(self, val_loss, model):
        if self.best_loss is None:
            self.best_loss = val_loss
            self.save_checkpoint(model)
        elif val_loss < self.best_loss - self.min_delta:
            self.best_loss = val_loss
            self.counter = 0
            self.save_checkpoint(model)
        else:
            self.counter += 1

        if self.counter >= self.patience:
            if self.restore_best_weights:
                model.load_state_dict(self.best_weights)
            return True
        return False

    def save_checkpoint(self, model):
        """Save model checkpoint"""
        self.best_weights = model.state_dict().copy()


# ============================================================
# Model Summary Function
# ============================================================
def get_model_summary(model):
    """Get model parameter summary"""
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    
    # Estimate model size in MB (assuming float32)
    model_size_mb = total_params * 4 / (1024 * 1024)
    
    return {
        'total_params': total_params,
        'trainable_params': trainable_params,
        'model_size_mb': model_size_mb
    }


# ============================================================
# Fixed Advanced Evaluation Metrics
# ============================================================
def krippendorff_alpha(data, level_of_measurement='nominal'):
    """Calculate Krippendorff's Alpha for agreement - FIXED VERSION"""
    y_true, y_pred = data[0], data[1]
    
    if len(y_true) != len(y_pred):
        return 0.0
    
    n = len(y_true)
    if n < 2:
        return 0.0
    
    if level_of_measurement == 'nominal':
        # Count observed agreements
        observed_agreements = np.sum(y_true == y_pred)
        
        # Calculate expected agreements by chance
        unique_vals = np.unique(np.concatenate([y_true, y_pred]))
        expected_agreement = 0.0
        
        for val in unique_vals:
            # Marginal frequencies
            freq_true = np.sum(y_true == val) / n
            freq_pred = np.sum(y_pred == val) / n
            expected_agreement += freq_true * freq_pred
        
        # Calculate alpha
        observed_prop = observed_agreements / n
        
        if expected_agreement >= 1.0:
            return 1.0 if observed_prop == 1.0 else 0.0
        
        alpha = (observed_prop - expected_agreement) / (1 - expected_agreement)
        return max(0.0, min(1.0, alpha))  # Clamp between 0 and 1
    
    else:  # For ordinal/interval data
        # Simplified version for continuous data
        mae = np.mean(np.abs(y_true - y_pred))
        data_range = np.max(np.concatenate([y_true, y_pred])) - np.min(np.concatenate([y_true, y_pred]))
        
        if data_range == 0:
            return 1.0 if mae == 0 else 0.0
        
        # Normalize MAE by data range
        normalized_disagreement = mae / data_range
        alpha = 1.0 - 2 * normalized_disagreement  # Simple transformation
        return max(0.0, min(1.0, alpha))

def sign_agreement_metric(y_true, y_pred):
    """Sign Agreement Metric (SAGR) - penalizes incorrect sign alongside deviation"""
    same_sign = np.sign(y_true) == np.sign(y_pred)
    mae = np.mean(np.abs(y_true - y_pred))
    sign_penalty = np.mean(~same_sign) * 2.0
    sagr = 1.0 / (1.0 + mae + sign_penalty)
    return sagr

def concordance_correlation_coefficient(y_true, y_pred):
    """Concordance Correlation Coefficient (CCC)"""
    mean_true = np.mean(y_true)
    mean_pred = np.mean(y_pred)
    var_true = np.var(y_true)
    var_pred = np.var(y_pred)
    
    if len(y_true) > 1:
        r, _ = pearsonr(y_true, y_pred)
    else:
        r = 0
    
    numerator = 2 * r * np.sqrt(var_true) * np.sqrt(var_pred)
    denominator = var_true + var_pred + (mean_true - mean_pred)**2
    
    if denominator == 0:
        return 0
    
    ccc = numerator / denominator
    return ccc

def evaluate_model_comprehensive(model, data_loader, device, num_classes=8):
    """Comprehensive evaluation with all requested metrics"""
    model.eval()
    
    all_exp_preds = []
    all_exp_true = []
    all_exp_probs = []
    all_val_preds = []
    all_val_true = []
    all_aro_preds = []
    all_aro_true = []
    
    with torch.no_grad():
        for imgs, (exp, val, aro) in data_loader:
            imgs = imgs.to(device)
            exp_logits, val_pred, aro_pred = model(imgs)
            
            # Expression (categorical)
            exp_probs = torch.softmax(exp_logits, dim=1)
            _, exp_pred = torch.max(exp_logits, 1)
            
            all_exp_preds.extend(exp_pred.cpu().numpy())
            all_exp_true.extend(exp.numpy())
            all_exp_probs.extend(exp_probs.cpu().numpy())
            
            # Valence and Arousal (continuous)
            all_val_preds.extend(val_pred.cpu().numpy())
            all_val_true.extend(val.numpy())
            all_aro_preds.extend(aro_pred.cpu().numpy())
            all_aro_true.extend(aro.numpy())
    
    # Convert to numpy arrays
    all_exp_preds = np.array(all_exp_preds)
    all_exp_true = np.array(all_exp_true)
    all_exp_probs = np.array(all_exp_probs)
    all_val_preds = np.array(all_val_preds)
    all_val_true = np.array(all_val_true)
    all_aro_preds = np.array(all_aro_preds)
    all_aro_true = np.array(all_aro_true)
    
    # Categorical metrics (Expression)
    exp_metrics = {}
    exp_metrics['Accuracy'] = accuracy_score(all_exp_true, all_exp_preds)
    exp_metrics['F1-Score'] = f1_score(all_exp_true, all_exp_preds, average='macro')
    exp_metrics['Cohens Kappa'] = cohen_kappa_score(all_exp_true, all_exp_preds)
    exp_metrics['Krippendorfs Alpha'] = krippendorff_alpha([all_exp_true, all_exp_preds])
    
    try:
        exp_true_bin = label_binarize(all_exp_true, classes=range(num_classes))
        if exp_true_bin.shape[1] == 1:
            exp_metrics['AUC'] = roc_auc_score(all_exp_true, all_exp_probs[:, 1])
        else:
            exp_metrics['AUC'] = roc_auc_score(exp_true_bin, all_exp_probs, 
                                             multi_class='ovr', average='macro')
    except:
        exp_metrics['AUC'] = 0.0
    
    try:
        if exp_true_bin.shape[1] == 1:
            exp_metrics['AUC-PR'] = average_precision_score(all_exp_true, all_exp_probs[:, 1])
        else:
            exp_metrics['AUC-PR'] = average_precision_score(exp_true_bin, all_exp_probs, average='macro')
    except:
        exp_metrics['AUC-PR'] = 0.0
    
    # Continuous metrics (Valence)
    val_metrics = {}
    val_metrics['RMSE'] = np.sqrt(np.mean((all_val_true - all_val_preds)**2))
    if len(all_val_true) > 1:
        val_corr, _ = pearsonr(all_val_true, all_val_preds)
        val_metrics['CORR'] = val_corr if not np.isnan(val_corr) else 0.0
    else:
        val_metrics['CORR'] = 0.0
    val_metrics['SAGR'] = sign_agreement_metric(all_val_true, all_val_preds)
    val_metrics['CCC'] = concordance_correlation_coefficient(all_val_true, all_val_preds)
    
    # Continuous metrics (Arousal)
    aro_metrics = {}
    aro_metrics['RMSE'] = np.sqrt(np.mean((all_aro_true - all_aro_preds)**2))
    if len(all_aro_true) > 1:
        aro_corr, _ = pearsonr(all_aro_true, all_aro_preds)
        aro_metrics['CORR'] = aro_corr if not np.isnan(aro_corr) else 0.0
    else:
        aro_metrics['CORR'] = 0.0
    aro_metrics['SAGR'] = sign_agreement_metric(all_aro_true, all_aro_preds)
    aro_metrics['CCC'] = concordance_correlation_coefficient(all_aro_true, all_aro_preds)
    
    return {
        'expression': exp_metrics,
        'valence': val_metrics,
        'arousal': aro_metrics,
        'predictions': {
            'exp_true': all_exp_true,
            'exp_pred': all_exp_preds,
            'val_true': all_val_true,
            'val_pred': all_val_preds,
            'aro_true': all_aro_true,
            'aro_pred': all_aro_preds
        }
    }

def print_evaluation_results(results, model_name, dataset_type="Validation"):
    """Print formatted evaluation results"""
    print(f"\n{'='*60}")
    print(f"EVALUATION RESULTS FOR {model_name} - {dataset_type}")
    print(f"{'='*60}")
    
    # Expression (Categorical) Results
    print(f"\nExpression Classification Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*55}")
    exp_descriptions = {
        'Accuracy': 'Accuracy',
        'F1-Score': 'F1-Score',
        'Cohens Kappa': 'Statistical measure of agreement',
        'Krippendorfs Alpha': 'Statistical measure of agreement',
        'AUC': 'Area under ROC curve',
        'AUC-PR': 'Area under Precision-Recall curve'
    }
    
    for metric, value in results['expression'].items():
        desc = exp_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")
    
    # Valence (Continuous) Results
    print(f"\nValence Regression Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*70}")
    val_descriptions = {
        'RMSE': 'Root Mean Square Error',
        'CORR': 'Pearson Correlation',
        'SAGR': 'Sign Agreement Metric',
        'CCC': 'Concordance Correlation Coefficient'
    }
    
    for metric, value in results['valence'].items():
        desc = val_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")
    
    # Arousal (Continuous) Results
    print(f"\nArousal Regression Metrics:")
    print(f"{'Metric':<25} {'Value':<10} {'Description'}")
    print(f"{'-'*70}")
    
    for metric, value in results['arousal'].items():
        desc = val_descriptions.get(metric, '')
        print(f"{metric:<25} {value:<10.4f} {desc}")


# ============================================================
# Sample Visualization Function
# ============================================================
def visualize_sample_predictions(model, dataset, device, num_samples=8, 
                               expression_names=None):
    """Visualize sample predictions with ground truth"""
    if expression_names is None:
        expression_names = ['Neutral', 'Happiness', 'Sadness', 'Surprise', 
                          'Fear', 'Disgust', 'Anger', 'Contempt']
    
    model.eval()
    
    # Select random samples
    indices = random.sample(range(len(dataset)), min(num_samples, len(dataset)))
    
    fig, axes = plt.subplots(2, 4, figsize=(20, 10))
    axes = axes.ravel()
    
    with torch.no_grad():
        for i, idx in enumerate(indices):
            # Get sample for visualization
            img_viz, true_exp, true_val, true_aro, img_path = dataset.get_sample_for_viz(idx)
            
            # Get model prediction
            img_tensor, (exp_tensor, val_tensor, aro_tensor) = dataset[idx]
            img_batch = img_tensor.unsqueeze(0).to(device)
            
            exp_logits, val_pred, aro_pred = model(img_batch)
            exp_probs = torch.softmax(exp_logits, dim=1)
            _, exp_pred = torch.max(exp_logits, 1)
            
            exp_pred = exp_pred.cpu().item()
            val_pred = val_pred.cpu().item()
            aro_pred = aro_pred.cpu().item()
            exp_conf = exp_probs[0][exp_pred].cpu().item()
            
            # Display image
            img_np = img_viz.permute(1, 2, 0).numpy()
            axes[i].imshow(img_np)
            axes[i].axis('off')
            
            # Create title with predictions
            title = f"True: {expression_names[true_exp]}\n"
            title += f"Pred: {expression_names[exp_pred]} ({exp_conf:.2f})\n"
            title += f"Val: {true_val:.2f}→{val_pred:.2f}\n"
            title += f"Aro: {true_aro:.2f}→{aro_pred:.2f}"
            
            # Color code: green for correct, red for incorrect
            color = 'green' if exp_pred == true_exp else 'red'
            axes[i].set_title(title, fontsize=10, color=color)
    
    plt.tight_layout()
    plt.suptitle("Sample Predictions vs Ground Truth", fontsize=16, y=1.02)
    plt.show()


# ============================================================
# Enhanced Training with Better Learning Rate Schedule
# ============================================================
def train_model(model, train_loader, val_loader, device, epochs=50, lr=0.001, 
                model_name="Model", patience=15, weight_decay=1e-4):
    """Enhanced training with better learning dynamics for custom CNN"""
    ce_loss = nn.CrossEntropyLoss(label_smoothing=0.1)  # Label smoothing for regularization
    mse_loss = nn.MSELoss()

    # Optimizer selection based on model type
    if "Custom" in model_name:
        # Custom CNN benefits from Adam with specific parameters
        optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay, 
                              betas=(0.9, 0.999), eps=1e-8)
        # More aggressive learning rate schedule for custom CNN
        scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.7, patience=3, 
                                    min_lr=1e-6, verbose=True)
    else:
        # Transfer learning models use AdamW
        optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
        scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, verbose=True)
    
    # Early stopping with different patience for custom CNN
    early_stopping = EarlyStopping(patience=patience, min_delta=0.001)

    history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
    
    print(f"\nTraining {model_name}...")
    print(f"Initial Learning Rate: {lr}")
    print(f"Weight Decay: {weight_decay}")
    print(f"Early Stopping Patience: {patience}")
    
    best_val_acc = 0.0

    for epoch in range(epochs):
        # Training phase
        model.train()
        total_loss, correct, total = 0, 0, 0

        for imgs, (exp, val, aro) in train_loader:
            imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)

            optimizer.zero_grad()
            exp_logits, val_pred, aro_pred = model(imgs)

            # Multi-task loss with balanced weighting
            loss_exp = ce_loss(exp_logits, exp)
            loss_val = mse_loss(val_pred, val)
            loss_aro = mse_loss(aro_pred, aro)
            
            # Balanced loss weighting (expression is primary task)
            if "Custom" in model_name:
                # Custom CNN focuses more on expression classification
                loss = 3.0 * loss_exp + 0.5 * loss_val + 0.5 * loss_aro
            else:
                loss = 2.0 * loss_exp + loss_val + loss_aro

            loss.backward()
            
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            
            optimizer.step()

            total_loss += loss.item() * imgs.size(0)
            _, preds = torch.max(exp_logits, 1)
            correct += (preds == exp).sum().item()
            total += exp.size(0)

        train_loss = total_loss / total
        train_acc = correct / total

        # Validation phase
        model.eval()
        val_loss, val_correct, val_total = 0, 0, 0
        with torch.no_grad():
            for imgs, (exp, val, aro) in val_loader:
                imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)
                exp_logits, val_pred, aro_pred = model(imgs)

                loss_exp = ce_loss(exp_logits, exp)
                loss_val = mse_loss(val_pred, val)
                loss_aro = mse_loss(aro_pred, aro)
                
                if "Custom" in model_name:
                    loss = 3.0 * loss_exp + 0.5 * loss_val + 0.5 * loss_aro
                else:
                    loss = 2.0 * loss_exp + loss_val + loss_aro

                val_loss += loss.item() * imgs.size(0)
                _, preds = torch.max(exp_logits, 1)
                val_correct += (preds == exp).sum().item()
                val_total += exp.size(0)

        val_loss /= val_total
        val_acc = val_correct / val_total
        
        # Learning rate scheduling
        scheduler.step(val_loss)

        # Track best validation accuracy
        if val_acc > best_val_acc:
            best_val_acc = val_acc

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)

        # Print progress every 3 epochs or last epoch
        print(f"Epoch {epoch}/{epochs} | "
                  f"Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | "
                  f"Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.6f}")
        
        # Early stopping check
        if early_stopping(val_loss, model):
            print(f"Early stopping triggered at epoch {epoch+1}")
            break
      # Print progress every 3 epochs or last epoch
    print(f"Epoch {epochs}/{epochs} | "
                  f"Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | "
                  f"Val Loss: {val_loss:.4f} Acc: {val_acc:.4f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.6f}")
    print(f"{model_name} training completed! Best Val Acc: {best_val_acc:.4f}")
    return history


def plot_history(history, title="Training Curves"):
    """Enhanced plotting function"""
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.plot(history["train_loss"], label="Train Loss", linewidth=2)
    plt.plot(history["val_loss"], label="Val Loss", linewidth=2)
    plt.legend()
    plt.title("Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 3, 2)
    plt.plot(history["train_acc"], label="Train Acc", linewidth=2)
    plt.plot(history["val_acc"], label="Val Acc", linewidth=2)
    plt.legend()
    plt.title("Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 3, 3)
    # Overfitting indicator
    overfitting = np.array(history["train_acc"]) - np.array(history["val_acc"])
    plt.plot(overfitting, label="Overfitting Gap", color='red', linewidth=2)
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.legend()
    plt.title("Overfitting Monitor")
    plt.xlabel("Epoch")
    plt.ylabel("Train Acc - Val Acc")
    plt.grid(True, alpha=0.3)
    
    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()


# ============================================================
# Debugging and Analysis Functions
# ============================================================
def analyze_model_gradients(model, data_loader, device):
    """Analyze gradient flow in the model"""
    model.train()
    grad_norms = []
    
    for imgs, (exp, val, aro) in data_loader:
        imgs, exp, val, aro = imgs.to(device), exp.to(device), val.to(device), aro.to(device)
        
        model.zero_grad()
        exp_logits, val_pred, aro_pred = model(imgs)
        
        loss_exp = F.cross_entropy(exp_logits, exp)
        loss_exp.backward()
        
        total_norm = 0
        for p in model.parameters():
            if p.grad is not None:
                param_norm = p.grad.data.norm(2)
                total_norm += param_norm.item() ** 2
        total_norm = total_norm ** (1. / 2)
        grad_norms.append(total_norm)
        
        break  # Just check first batch
    
    return grad_norms[0] if grad_norms else 0


def check_model_outputs(model, data_loader, device):
    """Check model output distributions"""
    model.eval()
    exp_outputs = []
    val_outputs = []
    aro_outputs = []
    
    with torch.no_grad():
        for imgs, (exp, val, aro) in data_loader:
            imgs = imgs.to(device)
            exp_logits, val_pred, aro_pred = model(imgs)
            
            exp_outputs.extend(torch.softmax(exp_logits, dim=1).cpu().numpy())
            val_outputs.extend(val_pred.cpu().numpy())
            aro_outputs.extend(aro_pred.cpu().numpy())
            
            break  # Just check first batch
    
    exp_outputs = np.array(exp_outputs)
    val_outputs = np.array(val_outputs)
    aro_outputs = np.array(aro_outputs)
    
    print(f"Expression logits - Mean: {exp_outputs.mean():.4f}, Std: {exp_outputs.std():.4f}")
    print(f"Valence outputs - Mean: {val_outputs.mean():.4f}, Std: {val_outputs.std():.4f}")
    print(f"Arousal outputs - Mean: {aro_outputs.mean():.4f}, Std: {aro_outputs.std():.4f}")


# ============================================================
# Main Training Function with All Models
# ============================================================
def main():
    # Configuration
    IMAGES_DIR = "/kaggle/input/dl-data/Dataset/Dataset/images"
    ANNOTATIONS_DIR = "/kaggle/input/dl-data/Dataset/Dataset/annotations"
    IMG_SIZE = (224, 224)
    BATCH_SIZE = 32
    EPOCHS = 10 
    
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")
    
    # Load data with train/val/test split
    train_loader, val_loader, test_loader, train_ds, val_ds, test_ds = get_dataloaders(
        IMAGES_DIR, ANNOTATIONS_DIR, img_size=IMG_SIZE, batch_size=BATCH_SIZE)
    
    print(f"Dataset loaded:")
    print(f"   • Train: {len(train_ds)} samples")
    print(f"   • Validation: {len(val_ds)} samples") 
    print(f"   • Test: {len(test_ds)} samples")
    
    # Expression class names (adjust according to your dataset)
    expression_names = ['Neutral', 'Happiness', 'Sadness', 'Surprise', 
                       'Fear', 'Disgust', 'Anger', 'Contempt']
    
    # Store all results
    all_results = {}
    all_models = {}
    
    # ========================
    # BASELINE 1: ResNet50
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING RESNET50 MODEL")
    print(f"{'='*70}")
    
    model_resnet = MultiTaskModel("resnet50", num_classes=8, dropout_rate=0.3).to(device)
    resnet_summary = get_model_summary(model_resnet)
    print(f"ResNet50 Model Summary:")
    print(f"   • Total Parameters: {resnet_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {resnet_summary['trainable_params']:,}")
    print(f"   • Model Size: {resnet_summary['model_size_mb']:.1f} MB")
    
    hist_resnet = train_model(model_resnet, train_loader, val_loader, device, 
                             epochs=EPOCHS, model_name="ResNet50", patience=10,
                             lr=0.0001, weight_decay=1e-4)
    plot_history(hist_resnet, title="ResNet50 Training Curves")
    
    # Validation evaluation
    results_resnet_val = evaluate_model_comprehensive(model_resnet, val_loader, device)
    print_evaluation_results(results_resnet_val, "RESNET50", "Validation")
    
    # Test evaluation
    results_resnet_test = evaluate_model_comprehensive(model_resnet, test_loader, device)
    print_evaluation_results(results_resnet_test, "RESNET50", "Test")
    
    # Store results
    all_results['ResNet50'] = {
        'validation': results_resnet_val,
        'test': results_resnet_test,
        'history': hist_resnet,
        'model_summary': resnet_summary
    }
    all_models['ResNet50'] = model_resnet
    
    # ========================
    # BASELINE 2: MobileNetV2
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING MOBILENETV2 MODEL")
    print(f"{'='*70}")
    
    model_mobilenet = MultiTaskModel("mobilenet_v2", num_classes=8, dropout_rate=0.3).to(device)
    mobilenet_summary = get_model_summary(model_mobilenet)
    print(f"MobileNetV2 Model Summary:")
    print(f"   • Total Parameters: {mobilenet_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {mobilenet_summary['trainable_params']:,}")
    print(f"   • Model Size: {mobilenet_summary['model_size_mb']:.1f} MB")
    
    hist_mobilenet = train_model(model_mobilenet, train_loader, val_loader, device, 
                                epochs=EPOCHS, model_name="MobileNetV2", patience=10,
                                lr=0.0001, weight_decay=1e-4)
    plot_history(hist_mobilenet, title="MobileNetV2 Training Curves")
    
    # Validation evaluation
    results_mobilenet_val = evaluate_model_comprehensive(model_mobilenet, val_loader, device)
    print_evaluation_results(results_mobilenet_val, "MOBILENETV2", "Validation")
    
    # Test evaluation
    results_mobilenet_test = evaluate_model_comprehensive(model_mobilenet, test_loader, device)
    print_evaluation_results(results_mobilenet_test, "MOBILENETV2", "Test")
    
    # Store results
    all_results['MobileNetV2'] = {
        'validation': results_mobilenet_val,
        'test': results_mobilenet_test,
        'history': hist_mobilenet,
        'model_summary': mobilenet_summary
    }
    all_models['MobileNetV2'] = model_mobilenet
    
    # ========================
    # IMPROVED CUSTOM CNN MODEL
    # ========================
    print(f"\n{'='*70}")
    print("TRAINING IMPROVED CUSTOM CNN MODEL")
    print(f"{'='*70}")
    
    model_custom = CustomCNN(num_classes=8, dropout_rate=0.3).to(device)
    custom_summary = get_model_summary(model_custom)
    print(f"Custom CNN Model Summary:")
    print(f"   • Total Parameters: {custom_summary['total_params']:,}")
    print(f"   • Trainable Parameters: {custom_summary['trainable_params']:,}")
    print(f"   • Model Size: {custom_summary['model_size_mb']:.1f} MB")
    
    # Debug model before training
    print(f"\nDEBUG: Checking model outputs before training...")
    check_model_outputs(model_custom, train_loader, device)
    grad_norm = analyze_model_gradients(model_custom, train_loader, device)
    print(f"Initial gradient norm: {grad_norm:.4f}")
    
    # Custom CNN training with optimized parameters
    hist_custom = train_model(model_custom, train_loader, val_loader, device, 
                             epochs=EPOCHS, lr=0.0000001, model_name="Custom CNN", 
                             patience=15)
    plot_history(hist_custom, title="Custom CNN Training Curves")
    
    # Validation evaluation
    results_custom_val = evaluate_model_comprehensive(model_custom, val_loader, device)
    print_evaluation_results(results_custom_val, "CUSTOM CNN", "Validation")
    
    # Test evaluation
    results_custom_test = evaluate_model_comprehensive(model_custom, test_loader, device)
    print_evaluation_results(results_custom_test, "CUSTOM CNN", "Test")
    
    # Store results
    all_results['Custom CNN'] = {
        'validation': results_custom_val,
        'test': results_custom_test,
        'history': hist_custom,
        'model_summary': custom_summary
    }
    all_models['Custom CNN'] = model_custom
    
    # Visualize sample predictions for each model
    for model_name, model in all_models.items():
        print(f"\nSAMPLE PREDICTIONS - {model_name.upper()}:")
        visualize_sample_predictions(model, test_ds, device, 
                                    expression_names=expression_names)
    
    # ========================
    # COMPREHENSIVE MODEL COMPARISON
    # ========================
    print(f"\n{'='*80}")
    print("COMPREHENSIVE MODEL COMPARISON")
    print(f"{'='*80}")
    
    # Create comparison tables
    models = ['ResNet50', 'MobileNetV2', 'Custom CNN']
    
    # Model Architecture Comparison
    print(f"\nMODEL ARCHITECTURE COMPARISON:")
    print(f"{'Model':<15} {'Parameters':<15} {'Size (MB)':<12} {'Type'}")
    print(f"{'-'*65}")
    for model_name in models:
        summary = all_results[model_name]['model_summary']
        model_type = "Transfer Learning" if model_name != "Custom CNN" else "Custom CNN"
        print(f"{model_name:<15} {summary['total_params']:>14,} {summary['model_size_mb']:>11.1f} {model_type}")
    
    # Validation Results Comparison
    print(f"\nVALIDATION RESULTS COMPARISON:")
    val_comparison_df = pd.DataFrame({
        model: [
            all_results[model]['validation']['expression']['Accuracy'],
            all_results[model]['validation']['expression']['F1-Score'],
            all_results[model]['validation']['expression']['AUC'],
            all_results[model]['validation']['valence']['RMSE'],
            all_results[model]['validation']['valence']['CORR'],
            all_results[model]['validation']['valence']['CCC'],
            all_results[model]['validation']['arousal']['RMSE'],
            all_results[model]['validation']['arousal']['CORR'],
            all_results[model]['validation']['arousal']['CCC']
        ] for model in models
    }, index=[
        'Exp_Accuracy', 'Exp_F1', 'Exp_AUC',
        'Val_RMSE', 'Val_CORR', 'Val_CCC',
        'Aro_RMSE', 'Aro_CORR', 'Aro_CCC'
    ])
    
    print(val_comparison_df.round(4))
    
    # Test Results Comparison
    print(f"\nTEST RESULTS COMPARISON:")
    test_comparison_df = pd.DataFrame({
        model: [
            all_results[model]['test']['expression']['Accuracy'],
            all_results[model]['test']['expression']['F1-Score'],
            all_results[model]['test']['expression']['AUC'],
            all_results[model]['test']['valence']['RMSE'],
            all_results[model]['test']['valence']['CORR'],
            all_results[model]['test']['valence']['CCC'],
            all_results[model]['test']['arousal']['RMSE'],
            all_results[model]['test']['arousal']['CORR'],
            all_results[model]['test']['arousal']['CCC']
        ] for model in models
    }, index=[
        'Exp_Accuracy', 'Exp_F1', 'Exp_AUC',
        'Val_RMSE', 'Val_CORR', 'Val_CCC',
        'Aro_RMSE', 'Aro_CORR', 'Aro_CCC'
    ])
    
    print(test_comparison_df.round(4))
    
    # Best model identification
    test_accuracies = {model: all_results[model]['test']['expression']['Accuracy'] 
                      for model in models}
    best_model = max(test_accuracies, key=test_accuracies.get)
    
    print(f"\nBEST MODEL: {best_model}")
    print(f"   • Test Accuracy: {test_accuracies[best_model]:.4f}")
    print(f"   • Model Size: {all_results[best_model]['model_summary']['model_size_mb']:.1f} MB")
    print(f"   • Parameters: {all_results[best_model]['model_summary']['total_params']:,}")
    
    # Training curves comparison
    plt.figure(figsize=(18, 12))
    
    # Plot training losses
    plt.subplot(3, 2, 1)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['train_loss'], label=f'{model_name} Train', linewidth=2)
    plt.title('Training Loss Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot validation losses
    plt.subplot(3, 2, 2)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['val_loss'], label=f'{model_name} Val', linewidth=2)
    plt.title('Validation Loss Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot training accuracies
    plt.subplot(3, 2, 3)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['train_acc'], label=f'{model_name} Train', linewidth=2)
    plt.title('Training Accuracy Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot validation accuracies
    plt.subplot(3, 2, 4)
    for model_name in models:
        hist = all_results[model_name]['history']
        plt.plot(hist['val_acc'], label=f'{model_name} Val', linewidth=2)
    plt.title('Validation Accuracy Comparison')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Plot overfitting comparison
    plt.subplot(3, 2, 5)
    for model_name in models:
        hist = all_results[model_name]['history']
        overfitting = np.array(hist['train_acc']) - np.array(hist['val_acc'])
        plt.plot(overfitting, label=f'{model_name}', linewidth=2)
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.title('Overfitting Comparison (Train - Val Accuracy)')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy Difference')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Performance vs efficiency scatter
    plt.subplot(3, 2, 6)
    for model_name in models:
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        size_mb = all_results[model_name]['model_summary']['model_size_mb']
        plt.scatter(size_mb, test_acc, s=200, label=model_name, alpha=0.7)
        plt.text(size_mb, test_acc, model_name, fontsize=9, ha='center', va='bottom')
    plt.title('Performance vs Model Size')
    plt.xlabel('Model Size (MB)')
    plt.ylabel('Test Accuracy')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.suptitle('Comprehensive Model Analysis', fontsize=16, y=0.98)
    plt.show()
    
    # Performance vs Efficiency Analysis
    print(f"\nPERFORMANCE VS EFFICIENCY ANALYSIS:")
    print(f"{'Model':<15} {'Test Acc':<10} {'Size (MB)':<12} {'Params (M)':<12} {'Efficiency'}")
    print(f"{'-'*75}")
    for model_name in models:
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        size_mb = all_results[model_name]['model_summary']['model_size_mb']
        params_m = all_results[model_name]['model_summary']['total_params'] / 1e6
        efficiency = test_acc / size_mb  # Accuracy per MB
        print(f"{model_name:<15} {test_acc:<10.4f} {size_mb:<12.1f} {params_m:<12.1f} {efficiency:<10.4f}")
    
    # Overfitting Analysis
    print(f"\nOVERFITTING ANALYSIS:")
    print(f"{'Model':<15} {'Val Acc':<10} {'Test Acc':<10} {'Gap':<10} {'Status'}")
    print(f"{'-'*60}")
    for model_name in models:
        val_acc = all_results[model_name]['validation']['expression']['Accuracy']
        test_acc = all_results[model_name]['test']['expression']['Accuracy']
        gap = val_acc - test_acc
        status = "Overfit" if gap > 0.05 else "Good" if gap > -0.02 else "Underfit"
        print(f"{model_name:<15} {val_acc:<10.4f} {test_acc:<10.4f} {gap:<10.4f} {status}")
    
    # Final summary and recommendations
    print(f"\nKEY INSIGHTS:")
    
    # Find most efficient model
    efficiencies = {model: all_results[model]['test']['expression']['Accuracy'] / 
                   all_results[model]['model_summary']['model_size_mb'] 
                   for model in models}
    most_efficient = max(efficiencies, key=efficiencies.get)
    smallest_model = min(models, key=lambda x: all_results[x]['model_summary']['model_size_mb'])
    
    print(f"   • Best Overall Performance: {best_model}")
    print(f"   • Most Efficient Model: {most_efficient}")
    print(f"   • Smallest Model: {smallest_model}")
    
    # Custom CNN specific analysis
    custom_acc = all_results['Custom CNN']['test']['expression']['Accuracy']
    if custom_acc > 0.25:  # Better than random + some margin
        print(f"   • Custom CNN Performance: GOOD ({custom_acc:.4f})")
    else:
        print(f"   • Custom CNN Performance: NEEDS IMPROVEMENT ({custom_acc:.4f})")
    
    # Regularization effectiveness
    print(f"\nREGULARIZATION EFFECTIVENESS:")
    for model_name in models:
        hist = all_results[model_name]['history']
        final_gap = hist['train_acc'][-1] - hist['val_acc'][-1]
        print(f"   • {model_name}: Final overfitting gap = {final_gap:.4f}")
    
    return all_models, all_results

if __name__ == "__main__":
    models, results = main()
```
